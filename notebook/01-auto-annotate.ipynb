```python
import subprocess

def cmd(command_string):
    """Executes a raw string directly in the system shell."""
    result = subprocess.run(command_string, shell=True, capture_output=True, text=True)
    
    if result.returncode == 0:
        print(result.stdout)
    else:
        print("Error:\n", result.stderr)

# Usage:
# cmd("echo 'Hello World' > test.txt && cat test.txt")
```

```python
import shutil
from pathlib import Path

IMAGES_DIR = Path("dataset/images")
IMAGES2_DIR = Path("dataset/images2")

START_INDEX = 469
END_INDEX = 539

IMAGES2_DIR.mkdir(exist_ok=True)

image_files = sorted(IMAGES_DIR.glob("*.jpg"))

for i in range(START_INDEX - 1, END_INDEX):
    src = image_files[i]
    dst = IMAGES2_DIR / src.name
    shutil.copy2(src, dst)
    print(f"Copied: {src.name}")
```
